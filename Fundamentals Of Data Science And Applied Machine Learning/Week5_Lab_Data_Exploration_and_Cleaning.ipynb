```python
# เปลี่ยนข้อมูลใน column Pclass เป็น cat
new_Pclass = pd.Categorical(titanic_train["Pclass"], ordered=True)

new_Pclass = new_Pclass.rename_categories(["Class1","Class2","Class3"])

new_Pclass.describe()
```
